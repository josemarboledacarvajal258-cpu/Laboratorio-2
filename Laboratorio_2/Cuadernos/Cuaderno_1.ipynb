### Ejercicio 1: Sistema $3 \times 3$ (Regla de Cramer y Cofactores)

**Enunciado del sistema:**

$$
\begin{aligned} 
2x - y + 3z &= 10 \\ 
x + 4y - z &= -5 \\ 
3x - 2y + 5z &= 12 
\end{aligned}
$$

**Representación matricial $Ax = b$:**
$$
\begin{bmatrix} 
2 & -1 & 3 \\ 
1 & 4 & -1 \\ 
3 & -2 & 5 
\end{bmatrix} 
\begin{bmatrix} 
x \\ 
y \\ 
z 
\end{bmatrix} = 
\begin{bmatrix} 
10 \\ 
-5 \\ 
12 
\end{bmatrix}
$$

**Cálculo del determinante**

- $\det(A)$ por Cofactores

Fórmula de Cramer:
$$x_j = \frac{\det(A_j)}{\det(A)}$$

Desarrollo del determinante:
$$
\begin{aligned} 
\det(A) &= 2 \begin{vmatrix} 4 & -1 \\ -2 & 5 \end{vmatrix} - (-1) \begin{vmatrix} 1 & -1 \\ 3 & 5 \end{vmatrix} + 3 \begin{vmatrix} 1 & 4 \\ 3 & -2 \end{vmatrix} \\ 
&= 2(20 - 2) + 1(5 + 3) + 3(-2 - 12) \\ 
&= 2(18) + 1(8) + 3(-14) \\ 
&= 36 + 8 - 42 \\
&= 2 
\end{aligned}
$$

El determianate es distinto de 0, por lo tanto el sisitema tienne solución.

- **Valor de x:** Para hallar x, se reemplaza la primera columna de la matriz principal por el vector de términos independientes, se calcula su determinante y se divide por el determinante de la matriz principal:

$$x = \frac{\begin{vmatrix} 10 & -1 & 3 \\ -5 & 4 & -1 \\ 12 & -2 & 5 \end{vmatrix}}{2} = \frac{53}{2}$$

- **Valor de y:** Para hallar x, se reemplaza la segunda columna de la matriz principal por el vector de términos independientes, repitiendo el proceso para hallar x:

$$y = \frac{\begin{vmatrix} 2 & 10 & 3 \\ 1 & -5 & -1 \\ 3 & 12 & 5 \end{vmatrix}}{2} = -\frac{25}{2}$$

- **Valor de z:** Para halar z, se reemplaza la tercera columna de la matriz principal por el vector de términos independientes, repitiendo el proceso de los dos valores anteriores:

$$z = \frac{\begin{vmatrix} 2 & -1 & 10 \\ 1 & 4 & -5 \\ 3 & -2 & 12 \end{vmatrix}}{2} = \frac{-37}{2}$$

Por lo tanto, las soluciones del sisitema son:
$$
\begin{aligned}
x = \frac{53}{2}, y = -\frac{25}{2},  z = -\frac{37}{2}
\end{aligned}
$$

**Comprobación de las soluciones**
$$
\begin{aligned}
2\left(\frac{53}{2}\right) - \left(-\frac{25}{2}\right) + 3\left(-\frac{37}{2}\right) &= 10 \\[10pt]
53 + \frac{25}{2} - \frac{111}{2} &= 10 \\[10pt]
10 &= 10
\end{aligned}
$$

### Ejercicio 3: Sistema $2 \times 2$ (Método de sustitución)

**Enunciado del sistema**

$$
\begin{aligned}
2x + 3y &= 11 &(1)\\
4x + y &= 14 &(2)\\
\end{aligned}
$$

Para resolver este sistema 2x2, se pueden usar tres métodos distintos, sustitución, igualación o reducción, para este ejemplo, el sistema de ecuaciones se resolverá por el método de sustitución:

Despejo $x$ de 1:

$$
\begin{aligned}
2x + 3y &= 11 \\
x = \frac{11 - 3y}{2}
\end{aligned}
$$

Sustituyo $x$ en 2 y obtengo $y$:

$$
\begin{aligned}
4\left(\frac{11 - 3y}{2}\right) + y &= 14 \\[10pt]
2(11 - 3y) + y &= 14 \\[10pt]
22 - 6y &= 14 \\[10pt]
-6y &= 14 - 22 \\[10pt]
y &= -\frac{8}{-6} \\[10pt]
y &= \frac{4}{3}
\end{aligned}
$$

Sustituyo $y$ en 2 para hallar $x$:

$$
\begin{aligned}
2x + 3\left(\frac{4}{3}\right) &= 11  \\[10pt]
2x + 4 &= 11 \\[10pt]
2x &= 7 \\[10pt]
x &= \frac{7}{2}
\end{aligned}
$$

**Solución del sistema**

$$x = \frac{7}{2}, \quad y = \frac{4}{3}$$

**Comprobación de las soluciones**

$$
\begin{aligned}
2\left(\frac{7}{2}\right) + 3\left(\frac{4}{3}\right) &= 11 \\[10pt]
7 + 4 &= 11 \\[10pt]
11 &= 11
\end{aligned}
$$

### Análisis del Residuo en los Sistemas de Ecuaciones Lineales

El análisis del residuo es un procedimiento utilizado en los métodos numéricos para evaluar la calidad de las soluciones aproximadas de un sistema de ecuaciones lineales. Consiste en calcular la diferencia entre los términos independientes originales y los resultados obtenidos al sustituir las soluciones calculadas. Se expresa mediante:

$$
r = b - A\hat{x}
$$

Un residuo cercano a cero indica que la solución satisface adecuadamente las ecuaciones originales, mientras que un residuo elevado representa una mayor diferencia en los resultados.

**Ventajas:**
- Permite verificar la calidad de las soluciones obtenidas.
- Ayuda a identificar errores en los cálculos.
- Facilita el control de precisión.
- Permite establecer criterios de convergencia en métodos iterativos.
- Ayuda a comparar diferentes métodos numéricos.

**Aplicaciones:**
Se utiliza en métodos como Gauss-Seidel y Jacobi, en el análisis estructural, en simulaciones computacionales y en diferentes problemas de ingeniería donde se requiere comprobar la confiabilidad de los resultados numéricos.

**Conclusión:**
El análisis del residuo es una herramienta importante porque permite comprobar qué tan bien una solución aproximada satisface un sistema de ecuaciones, ayudando a obtener resultados más precisos y confiables en aplicaciones matemáticas y de ingeniería.
