Para que el código funciones, las celdas se deben ejecutar en orden (1, 2, 3 y 4)

```python
def pedir_numero(mensaje):
    while True:
        try:
            return float(input(mensaje).strip().replace(",", "."))
        except ValueError:
            print("  Valor no válido, intenta de nuevo.")


def pedir_tamano():
    while True:
        try:
            n = int(input("Número de ecuaciones (2, 3 o 4): "))
            if 2 <= n <= 4:
                return n
        except ValueError:
            pass
        print("  Ingresa un entero entre 2 y 4.")


def pedir_metodo():
    print("\nMétodo: 1) Newton-Raphson   2) Punto fijo   3) Comparar despejes")
    while True:
        op = input("Elige (1/2/3): ").strip()
        if op in ("1", "2", "3"):
            return {"1": "newton", "2": "punto_fijo", "3": "comparar"}[op]
        print("  Opción no válida.")


def pedir_expresiones(n, letra):
    return [input(f"  {letra}{i + 1}(x) = ") for i in range(n)]


def pedir_datos():
    n = pedir_tamano()
    metodo = pedir_metodo()

    if metodo == "newton":
        print("\nEcuaciones igualadas a cero, con x1, x2, ... (ej: x1**2 + x2**2 - 4)")
        expresiones = pedir_expresiones(n, "f")
    elif metodo == "punto_fijo":
        print("\nEcuaciones despejadas, solo la parte derecha (ej: (x1**2 + x2**2 + 8)/10)")
        expresiones = pedir_expresiones(n, "g")
    else:
        cantidad = int(pedir_numero("¿Cuántos despejes quieres comparar? "))
        expresiones = {}
        for d in range(1, cantidad + 1):
            print(f"\nDespeje {d}: solo la parte derecha de cada x_i = g_i(x)")
            expresiones[f"Despeje {d}"] = pedir_expresiones(n, "g")

    x0 = [pedir_numero(f"  Valor inicial de x{i + 1}: ") for i in range(n)]
    tol = pedir_numero("Tolerancia a cumplir (ej. 1e-8): ")

    return {"n": n, "metodo": metodo, "expresiones": expresiones, "x0": x0, "tol": tol}
```

```python
def ejecutar(datos):
    n, metodo, x0, tol = datos["n"], datos["metodo"], datos["x0"], datos["tol"]
    nombres = {"newton": "Newton-Raphson", "punto_fijo": "Punto fijo"}

    try:
        if metodo == "comparar":
            return comparar_despejes(datos["expresiones"], x0, tol=tol)

        F = crear_funcion(datos["expresiones"], n)
        if metodo == "newton":
            solucion, it, error, historial = newton_raphson(F, x0, tol=tol)
        else:
            solucion, it, error, historial = punto_fijo(F, x0, tol=tol)

    except ErrorConvergencia as e:
        print(f"\nError: {e}\n\nHistorial hasta el fallo:")
        mostrar_historial(e.historial)
        graficar_errores(e.historial, f"{nombres[metodo]} (no convergió)")
        return
    except Exception as e:
        print(f"\nError al evaluar las ecuaciones: {e}")
        return

    print(f"\nConvergió en {it} iteraciones (error = {error:.2e})\n")
    mostrar_historial(historial)
    print("\nSolución:")
    for i, v in enumerate(solucion, start=1):
        print(f"  x{i} = {v:.8f}")
    graficar_errores(historial, nombres[metodo])
```

```python
datos = pedir_datos()
ejecutar(datos)
```

### Ejercicio 1: Sistema $2 \times 2$ (Método de punto fijo)

**Enunciado del sistema**

$$
\begin{aligned}
y + x^2 - x &= \frac{1}{2} \\
x^2 + y^2 &= 1 
\end{aligned}
$$

Se proponen los despejes $x = \text{f}_1(x,y)$ y $y = \text{f}_2(x,y)$:

$$
\begin{aligned}
\text{f}_1(x,y) = \sqrt{x + \frac{1}{2} - y} \\
\text{f}_2(x,y = \sqrt{1-x^2}
\end{aligned}
$$

**Comprobación de convergencia:** $|\frac{\partial f_1}{\partial x}| + |\frac{\partial f_1}{dy}| < 1$

Derivadas parciales:

- Para $\text{f}_1(x,y)$:

$$
\begin{aligned}
\frac{\partial f_1}{\partial x} = \frac{1}{2\sqrt{x + \frac{1}{2} - y}} \\[10pt]
\frac{\partial f_1}{\partial y} = -\frac{1}{2\sqrt{x + \frac{1}{2} - y}}
\end{aligned}
$$

- Para $\text{f}_2(x,y)$:

$$
\begin{aligned}
\frac{\partial f_2}{\partial x} = -\frac{x}{\sqrt{1 - x^2}} \\[10pt]
\frac{\partial f_2}{\partial x} = 0
\end{aligned}
$$

Se propone el punto inicial $(0.8,0.5)$:

- Para $f_1(x,y)$:

$$
\left| \frac{1}{2\sqrt{(0.8) + \frac{1}{2}(0.5)}} \right| + \left| -\frac{1}{2\sqrt{(0.8) + \frac{1}{2}(0.5)}} \right| = 1.118 > 1 \text{ (No converge)}
$$

- Para $f_2(x,y)$:

$$
\left| -\frac{0.8}{\sqrt{1 - (0.8)^2}} \right| + |0| = 1.333 > 1 \text{ (No converge)}
$$

Como ambas sumas son mayores a 1 en el punto inicial $(0.8, 0.5)$, el criterio indica que este esquema localmente no garantiza convergencia. Aun así, realizamos las iteraciones requeridas para observar el comportamiento numérico.

**Iteración paso a paso:**

- **Iteración 1:**

$$
\begin{aligned}
x_1 &= f_1(0.8, 0.5) = \sqrt{0.8 + 0.5 - 0.5} = \sqrt{0.8} = 0.8944 \\
y_1 &= f_2(0.8, 0.5) = \sqrt{1 - (0.8)^2} = \sqrt{0.36} = 0.6
\end{aligned}
$$

Nuevo punto de iteración: $(0.8944, 0.6)$

- **Iteración 2:**

$$
\begin{aligned}
x_2 &= f_1(0.8944, 0.6) = \sqrt{0.8944 + 0.5 - 0.6} = \sqrt{0.7944} = 0.8913 \\
y_2 &= f_2(0.8944, 0.6) = \sqrt{1 - (0.8944)^2} = \sqrt{0.2001} = 0.4473
\end{aligned}
$$

Nuevo punto de iteración: $(0.8913, 0.4473)$

- **Iteración 3:**

$$
\begin{aligned}
x_3 &= f_1(0.8913, 0.4473) = \sqrt{0.8913 + 0.5 - 0.4473} = \sqrt{0.944} = 0.9716 \\
y_3 &= f_2(0.8913, 0.4473) = \sqrt{1 - (0.8913)^2} = \sqrt{0.1949} = 0.4414
\end{aligned}
$$

### Ejercicio 2: Sistema $2 \times 2$ (Método de Newthon-Raphson)

**Enunciado del sistema**

$$
\begin{aligned}
F_1(x,y) = x^2 + y^2 - 4 = 0 \\
F_2(x,y) = xy - 1 = 0
\end{aligned}
$$

Matriz Jacobiana $J(x,y)$ y vector de funciones $F(x,y)$:

$$
F(x,y) = \begin{pmatrix} x^2 + y^2 - 4 \\ xy - 1 \end{pmatrix}
$$

Derivadas parciales:

- $\frac{\partial F_1}{\partial x} = 2x; \quad \frac{\partial F_1}{\partial y} = 2y$
- $\frac{\partial F_2}{\partial x} = y; \quad \frac{\partial F_2}{\partial y} = x$

$$
J(x,y) = \begin{pmatrix} 2x & 2y \\ y & x \end{pmatrix}
$$

Se propone el punto inicial $(2,0.5)$

- Vector de funciones $F(2, 0.5)$:

$$
\begin{aligned}
F_1(2, 0.5) &= (2)^2 + (0.5)^2 - 4 = 4 + 0.25 - 4 = 0.25 \\
F_2(2, 0.5) &= (2)(0.5) - 1 = 1 - 1 = 0
\end{aligned}
$$

$$
F(2, 0.5) = \begin{pmatrix} 0.25 \\ 0 \end{pmatrix}
$$

Matriz Jacobiana $J(2, 0.5)$:

$$
J(2, 0.5) = \begin{pmatrix} 2(2) & 2(0.5) \\ 0.5 & 2 \end{pmatrix} = \begin{pmatrix} 4 & 1 \\ 0.5 & 2 \end{pmatrix}
$$

Inversa de la matriz Jacobiana $J^{-1}(2, 0.5)$:

- Determinante de $J$: $\det(J) = 8 - 0.5 = 7.5$

$$
J^{-1} = \frac{1}{7.5} \begin{pmatrix} 2 & -1 \\ -0.5 & 4 \end{pmatrix}
$$

**Iteración con Newton-Raphson:**

$$
\begin{pmatrix} x_1 \\ y_1 \end{pmatrix} = \begin{pmatrix} x_0 \\ y_0 \end{pmatrix} - J^{-1}(x_0, y_0) F(x_0, y_0)
$$

Multiplicación de $J^{-1}F$:

$$
\frac{1}{7.5} \begin{pmatrix} 2 & -1 \\ -0.5 & 4 \end{pmatrix} \begin{pmatrix} 0.25 \\ 0 \end{pmatrix} = \frac{1}{7.5} \begin{pmatrix} 2(0.25) + 0 \\ -0.5(0.25) + 0 \end{pmatrix} = \frac{1}{7.5} \begin{pmatrix} 0.5 \\ -0.125 \end{pmatrix} = \begin{pmatrix} 0.0667 \\ -0.0167 \end{pmatrix}
$$

Actualización del vector de variables:

$$
\begin{pmatrix} x_1 \\ y_1 \end{pmatrix} = \begin{pmatrix} 2 \\ 0.5 \end{pmatrix} - \begin{pmatrix} 0.0667 \\ -0.0167 \end{pmatrix} = \begin{pmatrix} 1.9333 \\ 0.5167 \end{pmatrix}
$$

### Ejercicio 3: Sistema $2 \times 2$ (Comparación de despejes)

**Enunciado del sistema**

Despeje 1:

$$
\begin{aligned}
f_1(x,y) = \sqrt{2 - y} \\
f_2(x,y) = \sqrt{1.5 - x} 
\end{aligned}
$$

Derivadas parciales evaluadas en $(1, 0.5)$:

- $\frac{\partial f_1}{\partial x} = 0$
- $\frac{\partial f_1}{\partial y} = -\frac{1}{2\sqrt{2-y}} \Rightarrow -\frac{1}{2\sqrt{1.5}} \approx -0.4082$ \\[10pt]
- $\frac{\partial f_2}{\partial x} = -\frac{1}{2\sqrt{1.5-x}} \Rightarrow -\frac{1}{2\sqrt{0.5}} \approx -0.7071$
- $\frac{\partial f_2}{\partial y} = 0$

Suma de fracciones parciales:

- $|0| + |-0.4082| = 0.4082 < 1 \text{ (Cumple)}$
- $|-0.7071| + |0| = 0.7071 < 1 \text{ (Cumple)}$

Como ambas sumas son menores a 1, el despeje 1 es convergente.

Despeje 2:

$$
\begin{aligned}
f_1(x,y) = 1.5 - y^2 \\
f_2(x,y) = 2 - x^2
\end{aligned}
$$

Derivadas parciales evaluadas en $(1, 0.5)$:

- $\frac{\partial f_1}{\partial x} = 0$
- $\frac{\partial f_1}{\partial y} = -2y \Rightarrow -2(0.5) = -1$
- $\frac{\partial f_2}{\partial x} = -2x \Rightarrow -2(1) = -2$
- $\frac{\partial f_2}{\partial y} = 0$

Suma de las fracciones parciales:

- $|0| + |-1| = 1 = 1 \text{ (No converge)}$
- $|-2| + |0| = 2 > 1 \text{ (No converge)}$

Ambas sumas no cumplen con el criterio de convergencia, por lo tanto, el despeje 2 no converge.





### Análisis de tolerancia y control de actualizaciones

Las tolerancias y el control de las actualizaciones son herramientas fundamentales en los métodos numéricos para resolver sistemas de ecuaciones no lineales. Las tolerancias permiten establecer un margen de error aceptable para determinar cuándo detener las iteraciones y considerar que se ha alcanzado una solución suficientemente precisa. Por su parte, el control de las actualizaciones permite supervisar los cambios de las variables en cada iteración, verificando si las aproximaciones convergen hacia una solución estable.

**Ventajas y aplicaciones:** Permiten mejorar la precisión y confiabilidad de los resultados, evitar cálculos innecesarios y detectar problemas de convergencia. Se aplican en métodos como Newton-Raphson y Newton multivariable, así como en problemas de ingeniería, análisis estructural, simulaciones y cálculos matemáticos que requieren resolver sistemas de ecuaciones no lineales.
